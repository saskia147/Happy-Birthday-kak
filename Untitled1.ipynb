```python

# ================================================================
# PROGRAM ALGORITMA DIJKSTRA
# STUDI KASUS: JARINGAN PENGANTARAN PAKET KOTA BUKITTINGGI
# SATU CELL GOOGLE COLAB
# ================================================================

# 1. IMPORT LIBRARY
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd
import heapq
import time
import math

from matplotlib.gridspec import GridSpec
from matplotlib.patches import FancyBboxPatch, Rectangle, Circle, FancyArrowPatch
from matplotlib.lines import Line2D
from IPython.display import display
from IPython.display import Markdown

# ================================================================
# 2. PENGATURAN DESAIN
# ================================================================

BG = "#FBE8D7"
CREAM = "#FFF3E5"
MAROON = "#64251F"
DARK = "#39231F"
BROWN = "#795044"
GOLD = "#F5CC75"
PINK = "#EFC2BE"
RED = "#E52B32"
BLUE = "#A8D8E8"
WHITE = "#FFFDF8"
GREEN = "#367D55"

plt.rcParams["font.family"] = "DejaVu Sans"
plt.rcParams["figure.facecolor"] = BG
plt.rcParams["axes.facecolor"] = CREAM

# ================================================================
# 3. DATA SIMPUL (VERTEX)
# ================================================================

locations = {
    "A": "J&T Express\nPasar Aur Kuning",
    "B": "Terminal\nSimpang Aur",
    "C": "Aur Kuning",
    "D": "Tarok Dipo",
    "E": "Garegeh",
    "F": "Kubu Tanjung",
    "G": "Pulai Anak Air",
    "H": "Jam Gadang",
    "I": "Bukit Apit Puhun",
    "J": "Panorama Baru",
    "K": "Titik K",
    "L": "Birugo",
    "M": "Pakan Kurai",
    "N": "Parit Antang",
    "O": "Koto Selayan"
}

# ================================================================
# 4. DATA SISI (EDGE) DAN BOBOT JARAK
# ================================================================

edges = [
    ("A", "B", 1.5),
    ("A", "C", 1.3),
    ("A", "D", 2.2),
    ("A", "G", 3.5),
    ("A", "I", 7.1),
    ("B", "D", 0.8),
    ("B", "C", 0.7),
    ("C", "I", 6.0),
    ("D", "E", 3.7),
    ("D", "G", 2.3),
    ("E", "F", 3.8),
    ("E", "H", 3.7),
    ("F", "H", 6.0),
    ("G", "F", 4.1),
    ("G", "H", 3.1),
    ("I", "J", 3.8),
    ("I", "N", 5.9),
    ("J", "K", 7.0),
    ("J", "N", 8.0),
    ("K", "L", 1.1),
    ("K", "M", 2.8),
    ("L", "M", 2.8),
    ("L", "O", 4.8),
    ("M", "O", 3.2),
    ("N", "O", 1.5)
]

# Membuat graf tidak berarah
G = nx.Graph()

for node, name in locations.items():
    G.add_node(node, name=name)

for u, v, weight in edges:
    G.add_edge(u, v, weight=weight)

# Membuat tabel
df_nodes = pd.DataFrame([
    {"Kode": k, "Nama Lokasi": v.replace("\n", " ")}
    for k, v in locations.items()
])

df_edges = pd.DataFrame(
    edges,
    columns=["Dari", "Ke", "Jarak (km)"]
)

# ================================================================
# 5. MENAMPILKAN DATA GRAF
# ================================================================

display(Markdown("## DATA SIMPUL (VERTEX)"))
display(df_nodes)

display(Markdown("## DATA HUBUNGAN DAN JARAK (EDGE)"))
display(df_edges)

V = G.number_of_nodes()
E = G.number_of_edges()

print("\nSTATISTIK GRAF")
print("------------------------------")
print("Jumlah simpul (V) :", V)
print("Jumlah sisi (E)   :", E)
print("Ukuran graf       :", V + E)
print("Jenis graf        : Tidak berarah")

# ================================================================
# 6. MENENTUKAN POSISI MANUAL SIMPUL
# ================================================================

pos = {
    "A": (0.0, 0.0),
    "B": (1.0, 2.5),
    "C": (-1.0, 1.3),
    "D": (3.0, 2.5),
    "E": (5.2, 2.5),
    "F": (7.5, 1.8),
    "G": (3.3, 0.4),
    "H": (6.3, -0.2),

    "I": (1.2, -2.0),
    "J": (3.3, -2.7),
    "K": (5.5, -2.7),
    "L": (8.0, -1.7),
    "M": (7.7, -3.5),
    "N": (4.0, -4.3),
    "O": (9.7, -3.9)
}

# Posisi label nama lokasi
name_offsets = {
    "A": (-48, -12),
    "B": (-34, 25),
    "C": (-40, -5),
    "D": (0, 27),
    "E": (0, 27),
    "F": (35, 7),
    "G": (32, 0),
    "H": (35, 5),
    "I": (0, -32),
    "J": (0, -30),
    "K": (0, -30),
    "L": (35, 12),
    "M": (35, -3),
    "N": (0, -32),
    "O": (35, 0)
}

# ================================================================
# 7. FUNGSI VISUALISASI GRAF
# ================================================================

def draw_network(ax, shortest_path=None, infographic=False):

    if shortest_path is None:
        shortest_path = []

    # Menggambar seluruh sisi
    nx.draw_networkx_edges(
        G,
        pos,
        ax=ax,
        edge_color=BROWN,
        width=1.7,
        alpha=0.95
    )

    # Warna setiap simpul
    colors = []

    for node in G.nodes():
        if node == "A":
            colors.append(GOLD)
        elif node == "O":
            colors.append("#F3B8A9")
        else:
            colors.append("#F9DE9D")

    # Menggambar simpul
    nx.draw_networkx_nodes(
        G,
        pos,
        ax=ax,
        node_color=colors,
        node_size=800 if infographic else 950,
        edgecolors=DARK,
        linewidths=2
    )

    # Label kode simpul
    nx.draw_networkx_labels(
        G,
        pos,
        ax=ax,
        font_size=10 if infographic else 12,
        font_weight="bold",
        font_color=DARK
    )

    # Label bobot setiap sisi
    for u, v, data in G.edges(data=True):

        x1, y1 = pos[u]
        x2, y2 = pos[v]

        mx = (x1 + x2) / 2
        my = (y1 + y2) / 2

        dx = x2 - x1
        dy = y2 - y1

        length = math.sqrt(dx**2 + dy**2)

        if length > 0:
            offset = 0.13 if infographic else 0.16
            mx += (-dy / length) * offset
            my += (dx / length) * offset

        weight = data["weight"]

        ax.text(
            mx,
            my,
            f"{weight:.1f} km".replace(".", ","),
            fontsize=7.5 if infographic else 9,
            ha="center",
            va="center",
            color=DARK,
            fontweight="medium",
            bbox=dict(
                boxstyle="round,pad=0.2",
                facecolor=CREAM,
                edgecolor="none",
                alpha=0.95
            ),
            zorder=5
        )

    # Label nama lokasi dalam kotak merah muda
    for node, (x, y) in pos.items():

        dx, dy = name_offsets[node]

        ax.annotate(
            locations[node],
            xy=(x, y),
            xytext=(dx, dy),
            textcoords="offset points",
            ha="center",
            va="center",
            fontsize=7.3 if infographic else 8.5,
            fontweight="medium",
            color=DARK,
            bbox=dict(
                boxstyle="round,pad=0.45",
                facecolor=PINK,
                edgecolor="#A66E68",
                linewidth=0.8,
                alpha=0.98
            ),
            zorder=7
        )

    # Menggambar jalur terpendek menggunakan panah merah
    if len(shortest_path) > 1:

        for u, v in zip(shortest_path[:-1], shortest_path[1:]):

            arrow = FancyArrowPatch(
                posA=pos[u],
                posB=pos[v],
                arrowstyle="-|>",
                mutation_scale=18,
                color=RED,
                linewidth=4,
                shrinkA=18,
                shrinkB=18,
                zorder=8
            )

            ax.add_patch(arrow)

    ax.set_xlim(-2.0, 11.2)
    ax.set_ylim(-5.2, 3.6)
    ax.set_aspect("equal")
    ax.axis("off")

# ================================================================
# 8. VISUALISASI GRAF AWAL
# ================================================================

print("\nMENAMPILKAN GRAF JARINGAN AWAL...")

fig, ax = plt.subplots(figsize=(15, 9))

draw_network(ax)

ax.set_title(
    "GRAF JARINGAN PENGANTARAN PAKET KOTA BUKITTINGGI",
    fontsize=17,
    fontweight="bold",
    color=MAROON,
    pad=20
)

plt.tight_layout()
plt.show()

# ================================================================
# 9. IMPLEMENTASI ALGORITMA DIJKSTRA
# ================================================================

def dijkstra(graph, start, target):

    # Inisialisasi jarak
    distances = {
        node: float("inf")
        for node in graph.nodes()
    }

    distances[start] = 0

    # Menyimpan predecessor
    previous = {
        node: None
        for node in graph.nodes()
    }

    # Priority queue
    pq = [(0, start)]

    visited = set()
    history = []
    iteration = 0

    start_time = time.perf_counter()

    while pq:

        current_distance, current_node = heapq.heappop(pq)

        # Mengabaikan data antrean yang sudah diproses
        if current_node in visited:
            continue

        visited.add(current_node)
        iteration += 1

        checked = []
        updates = []
        predecessors = []

        # Memeriksa tetangga
        for neighbor in sorted(graph.neighbors(current_node)):

            weight = graph[current_node][neighbor]["weight"]

            if neighbor in visited:
                checked.append(
                    f"{neighbor} (sudah tetap)"
                )
                continue

            new_distance = current_distance + weight

            checked.append(
                f"{neighbor} ({weight:g} km)"
            )

            if new_distance < distances[neighbor]:

                old_distance = distances[neighbor]

                distances[neighbor] = new_distance
                previous[neighbor] = current_node

                heapq.heappush(
                    pq,
                    (new_distance, neighbor)
                )

                old_text = (
                    "∞" if math.isinf(old_distance)
                    else f"{old_distance:.1f}"
                )

                updates.append(
                    f"{neighbor}: {old_text} → {new_distance:.1f}"
                )

                predecessors.append(
                    f"{neighbor} ← {current_node}"
                )

        history.append({
            "Iterasi": iteration,
            "Simpul Dipilih": current_node,
            "Jarak Sementara": current_distance,
            "Tetangga Diperiksa": ", ".join(checked),
            "Perubahan Jarak": (
                "; ".join(updates) if updates else "-"
            ),
            "Predecessor": (
                "; ".join(predecessors) if predecessors else "-"
            )
        })

    execution_time = time.perf_counter() - start_time

    # Rekonstruksi jalur
    path = []
    current = target

    if not math.isinf(distances[target]):

        while current is not None:
            path.append(current)
            current = previous[current]

        path.reverse()

    return distances, previous, path, history, execution_time

# Titik awal dan tujuan
start_node = "A"
target_node = "O"

distances, previous, shortest_path, history, execution_time = dijkstra(
    G,
    start_node,
    target_node
)

total_distance = distances[target_node]

# ================================================================
# 10. MENAMPILKAN PROSES ITERASI DIJKSTRA
# ================================================================

print("\nTABEL PROSES ITERASI ALGORITMA DIJKSTRA")

df_history = pd.DataFrame(history)

display(df_history)

# ================================================================
# 11. MENAMPILKAN HASIL RUTE TERPENDEK
# ================================================================

print("\n" + "=" * 60)
print("HASIL PENCARIAN RUTE TERPENDEK")
print("=" * 60)

print("Titik awal  :", start_node, "-", locations[start_node].replace("\n", " "))
print("Titik tujuan:", target_node, "-", locations[target_node].replace("\n", " "))

print("\nUrutan rute:")

for i, node in enumerate(shortest_path, 1):
    print(
        f"{i}. {node} - {locations[node].replace(chr(10), ' ')}"
    )

print("\nRUTE TERPENDEK:", " → ".join(shortest_path))

print(
    "TOTAL JARAK:",
    f"{total_distance:.1f}".replace(".", ","),
    "km"
)

print("=" * 60)

# ================================================================
# 12. VALIDASI HASIL MENGGUNAKAN NETWORKX
# ================================================================

nx_path = nx.dijkstra_path(
    G,
    source=start_node,
    target=target_node,
    weight="weight"
)

nx_distance = nx.dijkstra_path_length(
    G,
    source=start_node,
    target=target_node,
    weight="weight"
)

# Menghitung ulang total jarak dari sisi jalur
manual_distance = sum(
    G[u][v]["weight"]
    for u, v in zip(shortest_path[:-1], shortest_path[1:])
)

assert V == 15
assert E == 25
assert not G.is_directed()
assert shortest_path == ["A", "I", "N", "O"]
assert abs(total_distance - 14.5) < 1e-9
assert abs(nx_distance - total_distance) < 1e-9
assert abs(manual_distance - total_distance) < 1e-9

print("\nVALIDASI PROGRAM BERHASIL")
print("Jumlah simpul       :", V)
print("Jumlah sisi         :", E)
print("Rute Dijkstra       :", shortest_path)
print("Rute NetworkX       :", nx_path)
print("Total jarak         :", total_distance, "km")
print("Waktu eksekusi      :", f"{execution_time:.8f} detik")
print("Status validasi     : BERHASIL")

# ================================================================
# 13. VISUALISASI GRAF AKHIR
# ================================================================

print("\nVISUALISASI GRAF DENGAN RUTE TERPENDEK")

fig, ax = plt.subplots(figsize=(15, 9))

draw_network(
    ax,
    shortest_path=shortest_path
)

ax.set_title(
    "RUTE TERPENDEK ALGORITMA DIJKSTRA",
    fontsize=18,
    fontweight="bold",
    color=MAROON,
    pad=20
)

plt.tight_layout()
plt.show()

# ================================================================
# 14. MEMBUAT INFOGRAFIS FINAL
# ================================================================

fig = plt.figure(
    figsize=(20, 13.5),
    facecolor=BG
)

gs = GridSpec(
    2,
    2,
    figure=fig,
    width_ratios=[3.15, 1.12],
    height_ratios=[4.5, 1.25],
    left=0.035,
    right=0.975,
    top=0.865,
    bottom=0.055,
    wspace=0.09,
    hspace=0.15
)

# ------------------------------------------------
# HEADER UTAMA
# ------------------------------------------------

header = FancyBboxPatch(
    (0.23, 0.895),
    0.54,
    0.085,
    boxstyle="round,pad=0.01,rounding_size=0.025",
    facecolor=MAROON,
    edgecolor="#D99A59",
    linewidth=3,
    transform=fig.transFigure,
    zorder=10
)

fig.add_artist(header)

fig.text(
    0.5,
    0.938,
    "ALGORITMA DIJKSTRA",
    ha="center",
    va="center",
    fontsize=27,
    fontweight="bold",
    color=WHITE,
    zorder=11
)

fig.text(
    0.5,
    0.885,
    "Menentukan Rute Terpendek Kurir Paket di Kota Bukittinggi",
    ha="center",
    fontsize=14,
    fontweight="bold",
    color=DARK
)

# Ornamen paket sederhana pada header
package_ax = fig.add_axes([0.055, 0.89, 0.09, 0.09])
package_ax.set_xlim(0, 1)
package_ax.set_ylim(0, 1)
package_ax.axis("off")

package_ax.add_patch(
    Rectangle(
        (0.2, 0.12), 0.58, 0.58,
        facecolor="#C98B54",
        edgecolor=MAROON,
        linewidth=2
    )
)

package_ax.plot(
    [0.49, 0.49], [0.12, 0.70],
    color=MAROON, linewidth=2
)

package_ax.plot(
    [0.2, 0.49, 0.78],
    [0.7, 0.88, 0.7],
    color=MAROON,
    linewidth=2
)

# Ornamen pin lokasi
pin_ax = fig.add_axes([0.86, 0.89, 0.08, 0.09])
pin_ax.set_xlim(0, 1)
pin_ax.set_ylim(0, 1)
pin_ax.axis("off")

pin_ax.add_patch(
    Circle(
        (0.5, 0.65),
        0.25,
        facecolor=RED,
        edgecolor=WHITE,
        linewidth=2
    )
)

pin_ax.add_patch(
    Circle(
        (0.5, 0.65),
        0.09,
        facecolor=WHITE,
        edgecolor="none"
    )
)

pin_ax.plot(
    [0.35, 0.5, 0.65],
    [0.48, 0.08, 0.48],
    color=RED,
    linewidth=4
)

# ------------------------------------------------
# PANEL GRAF
# ------------------------------------------------

ax_graph = fig.add_subplot(gs[0, 0])

ax_graph.set_facecolor(CREAM)

ax_graph.add_patch(
    FancyBboxPatch(
        (0.0, 0.0),
        1.0,
        1.0,
        transform=ax_graph.transAxes,
        facecolor=CREAM,
        edgecolor="#B77D62",
        linewidth=2,
        boxstyle="round,pad=0.015",
        zorder=-5
    )
)

draw_network(
    ax_graph,
    shortest_path=shortest_path,
    infographic=True
)

ax_graph.set_title(
    "GRAF JARINGAN PENGANTARAN PAKET",
    fontsize=14,
    fontweight="bold",
    color=WHITE,
    pad=12,
    backgroundcolor=MAROON
)

# Legenda
legend_items = [
    Line2D(
        [0], [0],
        marker="o",
        color="none",
        markerfacecolor=GOLD,
        markeredgecolor=DARK,
        markersize=10,
        label="Gudang / titik awal"
    ),
    Line2D(
        [0], [0],
        marker="o",
        color="none",
        markerfacecolor="#F3B8A9",
        markeredgecolor=DARK,
        markersize=10,
        label="Tujuan"
    ),
    Line2D(
        [0], [0],
        marker="o",
        color="none",
        markerfacecolor="#F9DE9D",
        markeredgecolor=DARK,
        markersize=10,
        label="Titik lainnya"
    ),
    Line2D(
        [0], [0],
        color=BROWN,
        linewidth=2,
        label="Jalan / hubungan"
    ),
    Line2D(
        [0], [0],
        color=RED,
        linewidth=4,
        label="Rute terpendek"
    )
]

ax_graph.legend(
    handles=legend_items,
    loc="lower left",
    fontsize=8,
    frameon=True,
    facecolor=WHITE,
    edgecolor=BROWN,
    ncol=2
)

# ------------------------------------------------
# PANEL TABEL DATA JARAK
# ------------------------------------------------

ax_table = fig.add_subplot(gs[0, 1])
ax_table.axis("off")

ax_table.set_title(
    "TABEL DATA JARAK ANTARTITIK",
    fontsize=12,
    fontweight="bold",
    color=WHITE,
    backgroundcolor=MAROON,
    pad=12
)

table_data = [
    [
        i,
        u,
        v,
        f"{w:.1f}".replace(".", ",")
    ]
    for i, (u, v, w) in enumerate(edges, 1)
]

table = ax_table.table(
    cellText=table_data,
    colLabels=["No.", "Dari", "Ke", "Jarak (km)"],
    cellLoc="center",
    colLoc="center",
    colWidths=[0.13, 0.22, 0.22, 0.43],
    bbox=[0, 0, 1, 0.97]
)

table.auto_set_font_size(False)
table.set_fontsize(8.5)

for (row, col), cell in table.get_celld().items():

    cell.set_edgecolor("#B68C7C")
    cell.set_linewidth(0.7)

    if row == 0:
        cell.set_facecolor(MAROON)
        cell.set_text_props(
            color=WHITE,
            weight="bold"
        )
    else:
        cell.set_facecolor(
            "#F6DCD1" if row % 2 == 0 else "#FFF4E9"
        )
        cell.set_text_props(color=DARK)

# ------------------------------------------------
# PANEL HASIL PERHITUNGAN
# ------------------------------------------------

ax_result = fig.add_subplot(gs[1, 0])
ax_result.axis("off")

ax_result.add_patch(
    FancyBboxPatch(
        (0.005, 0.02),
        0.99,
        0.96,
        boxstyle="round,pad=0.015,rounding_size=0.025",
        facecolor="#F7D9C8",
        edgecolor="#A86A50",
        linewidth=2,
        transform=ax_result.transAxes
    )
)

ax_result.text(
    0.04, 0.82,
    "HASIL PERHITUNGAN DIJKSTRA",
    fontsize=14,
    fontweight="bold",
    color=WHITE,
    bbox=dict(
        boxstyle="round,pad=0.5",
        facecolor=MAROON,
        edgecolor="none"
    ),
    transform=ax_result.transAxes
)

ax_result.text(
    0.05, 0.57,
    "Rute Terpendek:",
    fontsize=12,
    fontweight="bold",
    color=DARK,
    transform=ax_result.transAxes
)

ax_result.text(
    0.33, 0.40,
    " → ".join(shortest_path),
    fontsize=22,
    fontweight="bold",
    color=WHITE,
    ha="center",
    bbox=dict(
        boxstyle="round,pad=0.55",
        facecolor=MAROON,
        edgecolor="#D9A26C"
    ),
    transform=ax_result.transAxes
)

route_names = " → ".join(
    locations[node].replace("\n", " ")
    for node in shortest_path
)

ax_result.text(
    0.33, 0.19,
    route_names,
    fontsize=9,
    color=DARK,
    ha="center",
    transform=ax_result.transAxes
)

ax_result.text(
    0.73, 0.58,
    "TOTAL JARAK",
    fontsize=12,
    fontweight="bold",
    color=DARK,
    ha="center",
    transform=ax_result.transAxes
)

ax_result.text(
    0.73, 0.32,
    f"{total_distance:.1f}".replace(".", ",") + " km",
    fontsize=25,
    fontweight="bold",
    color=WHITE,
    ha="center",
    bbox=dict(
        boxstyle="round,pad=0.5",
        facecolor=MAROON,
        edgecolor="#D9A26C"
    ),
    transform=ax_result.transAxes
)

# ------------------------------------------------
# PANEL KESIMPULAN DAN KOMPLEKSITAS
# ------------------------------------------------

ax_conclusion = fig.add_subplot(gs[1, 1])
ax_conclusion.axis("off")

ax_conclusion.add_patch(
    FancyBboxPatch(
        (0.005, 0.02),
        0.99,
        0.96,
        boxstyle="round,pad=0.015,rounding_size=0.025",
        facecolor="#F7D9C8",
        edgecolor="#A86A50",
        linewidth=2,
        transform=ax_conclusion.transAxes
    )
)

ax_conclusion.text(
    0.06, 0.83,
    "KESIMPULAN",
    fontsize=13,
    fontweight="bold",
    color=WHITE,
    bbox=dict(
        boxstyle="round,pad=0.45",
        facecolor=MAROON,
        edgecolor="none"
    ),
    transform=ax_conclusion.transAxes
)

conclusion = (
    "Algoritma Dijkstra berhasil menentukan "
    "rute terpendek dari J&T Express Pasar Aur "
    "Kuning menuju Koto Selayan melalui "
    "Bukit Apit Puhun dan Parit Antang "
    f"dengan total jarak {total_distance:.1f}".replace(".", ",")
    + " km."
)

ax_conclusion.text(
    0.06, 0.65,
    conclusion,
    fontsize=9.2,
    color=DARK,
    va="top",
    wrap=True,
    linespacing=1.5,
    transform=ax_conclusion.transAxes
)

# Statistik ringkas pada bagian bawah
stats = (
    f"V = {V}  |  E = {E}  |  V + E = {V+E}\n"
    f"Waktu: {execution_time:.6f} detik\n"
    "Waktu: O((V+E) log V)\n"
    "Ruang: O(V+E)"
)

ax_conclusion.text(
    0.06, 0.28,
    stats,
    fontsize=8,
    color=MAROON,
    fontweight="medium",
    linespacing=1.4,
    transform=ax_conclusion.transAxes
)

# ------------------------------------------------
# 15. SIMPAN INFOGRAFIS
# ------------------------------------------------

filename = "Infografis_Algoritma_Dijkstra_Bukittinggi.png"

plt.savefig(
    filename,
    dpi=300,
    bbox_inches="tight",
    facecolor=BG
)

plt.show()

# ================================================================
# 16. VALIDASI AKHIR DAN DOWNLOAD
# ================================================================

print("\n" + "=" * 60)
print("PROGRAM SELESAI")
print("=" * 60)

print("Jumlah simpul        :", V)
print("Jumlah sisi          :", E)
print("Ukuran graf          :", V + E)
print("Jumlah iterasi       :", len(history))
print("Rute terpendek       :", " → ".join(shortest_path))
print("Total jarak          :", total_distance, "km")
print("Waktu eksekusi       :", f"{execution_time:.8f} detik")
print("Kompleksitas waktu   : O((V + E) log V)")
print("Kompleksitas ruang   : O(V + E)")
print("Validasi             : BERHASIL")
print("Nama file            :", filename)

print("=" * 60)

# Download otomatis jika dijalankan melalui Google Colab
try:
    from google.colab import files
    files.download(filename)
except ImportError:
    print("Download otomatis hanya tersedia di Google Colab.")
    print("File tersimpan di:", filename)
```
